### Storing image in matrix and converting them in Dark Mode

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
array1 = np.array([[1, 2, 3], [4, 5, 6]])
array2 = np.random.rand(3, 3)
array3 = np.zeros((4, 4))
```

```python
np.save('array1.npy', array1)
np.save('array2.npy', array2)
np.save('array3.npy', array3)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

# 1. Figure Setup
fig, ax = plt.subplots(figsize=(6, 6), dpi=150)
ax.set_facecolor('white')

# Colors matching the official 3D NumPy logo
blue_left = '#3B64C5'    # Dark Blue for 'N' face
cyan_top = '#4BAEE3'     # Cyan for top grid
cyan_right = '#2D9CDB'   # Bright Cyan for right grid

# Helper function to draw isometric polygons
def draw_poly(pts, color):
    poly = Polygon(pts, facecolor=color, edgecolor='white', linewidth=2)
    ax.add_patch(poly)

# --- TOP FACE (Cyan 2x2 grid) ---
draw_poly([[0, 2], [0.8, 2.4], [1.5, 2.05], [0.7, 1.65]], cyan_top)
draw_poly([[0.8, 2.4], [1.6, 2.8], [2.3, 2.45], [1.5, 2.05]], cyan_top)
draw_poly([[0.7, 1.65], [1.5, 2.05], [2.2, 1.7], [1.4, 1.3]], cyan_top)
draw_poly([[1.5, 2.05], [2.3, 2.45], [3.0, 2.1], [2.2, 1.7]], cyan_top)

# --- RIGHT FACE (Cyan 2x2 grid) ---
draw_poly([[1.4, 1.3], [2.2, 1.7], [2.2, 0.8], [1.4, 0.4]], cyan_right)
draw_poly([[2.2, 1.7], [3.0, 2.1], [3.0, 1.2], [2.2, 0.8]], cyan_right)
draw_poly([[1.4, 0.4], [2.2, 0.8], [2.2, -0.1], [1.4, -0.5]], cyan_right)
draw_poly([[2.2, 0.8], [3.0, 1.2], [3.0, 0.3], [2.2, -0.1]], cyan_right)

# --- LEFT FACE ('N' Shape in Dark Blue) ---
# Outer Left Face boundary forming 'N'
n_points = [
    [0, 2], [0.7, 1.65], [0.7, 0.85], [1.4, 1.3], [1.4, -0.5], 
    [0.7, -0.1], [0.7, 0.7], [0, 0.2]
]
draw_poly(n_points, blue_left)

# Add "NumPy" Text below logo
ax.text(1.5, -1.2, 'NumPy', fontsize=42, fontweight='bold', color=blue_left,
        ha='center', va='center', fontfamily='sans-serif')

ax.set_xlim(-0.3, 3.3)
ax.set_ylim(-1.6, 3.0)
plt.axis('off')

# 2. Convert Canvas to NumPy Array
fig.canvas.draw()
rgba_buffer = fig.canvas.buffer_rgba()
logo_array = np.asarray(rgba_buffer)[:, :, :3] # Keep RGB channels
plt.close(fig)

# 3. Save as .npy file
np.save("numpy_logo.npy", logo_array)
print("✅ Exact NumPy Logo successfully saved to 'numpy_logo.npy'!")

# 4. Display loaded array
plt.figure(figsize=(5, 5))
plt.imshow(logo_array)
plt.axis('off')
plt.show()
```
