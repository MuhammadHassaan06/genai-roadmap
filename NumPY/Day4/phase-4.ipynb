```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

# 1. Canvas Setup
fig, ax = plt.subplots(figsize=(6, 6), dpi=120)
ax.set_facecolor('white')

# 2. NumPy Official Theme Colors
navy_blue = '#013243'
sky_blue = '#4B9CD3'

# 3. Create Clean Grid Layout for NumPy Logo
# Top-Left Cube
ax.add_patch(Rectangle((0.1, 0.55), 0.35, 0.35, facecolor=navy_blue))
ax.add_patch(Rectangle((0.1, 0.55), 0.17, 0.17, facecolor=sky_blue))

# Top-Right Cube
ax.add_patch(Rectangle((0.55, 0.55), 0.35, 0.35, facecolor=sky_blue))
ax.add_patch(Rectangle((0.73, 0.55), 0.17, 0.17, facecolor=navy_blue))

# Bottom-Left Cube
ax.add_patch(Rectangle((0.1, 0.1), 0.35, 0.35, facecolor=sky_blue))
ax.add_patch(Rectangle((0.1, 0.1), 0.17, 0.17, facecolor=navy_blue))

# Bottom-Right Cube
ax.add_patch(Rectangle((0.55, 0.1), 0.35, 0.35, facecolor=navy_blue))
ax.add_patch(Rectangle((0.73, 0.1), 0.17, 0.17, facecolor=sky_blue))

# Center 'N' Text
ax.text(0.5, 0.5, 'N', fontsize=55, fontweight='bold', color=navy_blue, 
        ha='center', va='center', fontfamily='sans-serif')

ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
plt.axis('off')

# 4. Convert Canvas to NumPy Array (Matplotlib 3.8+ Compatible)
fig.canvas.draw()
rgba_buffer = fig.canvas.buffer_rgba()
logo_array = np.asarray(rgba_buffer)[:, :, :3] # Only RGB channels
plt.close(fig)

# 5. Save as .npy file
np.save("numpy_logo.npy", logo_array)
print("✅ High-Quality NumPy Logo saved successfully to 'numpy_logo.npy'!")

# 6. Display Array Output
plt.figure(figsize=(5, 5))
plt.imshow(logo_array)
plt.title("NumPy Logo Array")
plt.axis('off')
plt.show()
```
