```python
# Import the necessary libraries
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = Image.open("/kaggle/input/numpy-logo/numpy logo(github).png")
print(type(img))
print(img)

# summarize some details about the image
print(img.format)
print(img.size)
print(img.mode)
```

```python
img_np_array = np.array(img)
print(type(img_np_array))
print(img_np_array.shape)
```

```python
# change from RGBA to RGB
rgb_array = np.delete(arr=img_np_array, obj=3, axis=2)
print(rgb_array.shape)
```

```python
# check does image change ?
plt.imshow(rgb_array)
plt.show()
```

```python
with open('numpy logo.npy', 'wb') as logo_file:
  np.save(logo_file, rgb_array)
```

```python
with open('/kaggle/working/numpy logo.npy', 'rb') as logo_file:
  logo_array = np.load(logo_file)
print(logo_array.shape)
plt.imshow(logo_array)
plt.show()
```
